### Oefening 4
Stel $z = \sqrt{2} e^{3j\pi/5}$. Bepaal het reëel en imaginair deel. Bepaal (in graden) het argument (=poolhoek) van het complex toegevoegde van $(z+1)$.

```python
z=math.sqrt(2)*cmath.exp(3j*math.pi/5)
print("Het reëel deel van z is: ", z.real, "en het imaginair deel is: ", z.imag)

z1 = (z+1).conjugate()
print("De poolhoek is: ", math.degrees(cmath.phase(z1)))
```

### Oefening 8
Bereken m.b.v. vectoren de oppervlakte van de driehoek met hoekpunten $a(1, 2, 0)$, $b(3, 0, -3)$ en $c(5, 2, 6)$.

```python
a = np.array([1,2,0])
b = np.array([3,0,-3])
c = np.array([5,2,6])

ab = b-a
ac = c-a

print(1/2*np.linalg.norm(np.cross(ab,ac)))
```
