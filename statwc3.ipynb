### Oefening 20
Schets de kromme $y = \text{Bgtg} \left( \frac{x-1}{2} - \frac{1}{2x-2} \right)$. Bepaal de limiet in 1 en in $\pm\infty$.Bepaal $y'$. Zoek punten waarin $y' = 2$. Verklaar de resultaten op de grafiek.

```python
x = sp.symbols('x')
f = sp.atan((x-1)/2-1/(2*x-2))
sp.plotting.plot(f,(x,-10,10))
```

```python
lim1 = sp.limit(f,x,1)
print("De limiet van f voor x->1 is gelijk aan ", lim1)
lim1L=sp.limit(f,x,1,dir="-")
print("Voor x->1 is de linkerlimiet van f gelijk aan ",lim1L)
lim1R = sp.limit(f,x,1,dir='+')
print("Voor x->1 is de rechterlimiet van f gelijk aan ", lim1R)

limL=sp.limit(f,x,-sp.oo)
print("De limiet van f voor x->-oneindig is gelijk aan ", limL)
limR=sp.limit(f,x,+sp.oo)
print("De limiet van f voor x->+oneindig is gelijk aan ", limR)

afgeleide = sp.diff(f,x)
print("Afgeleide = ", afgeleide)
vgl = sp.Eq(afgeleide,2)
opl = sp.solve(vgl)
print("Opl = ", opl)

opln = sp.solve(vgl,0.5)
print("opln = ", opln)
```

### Oefening 23
Schets de krommen $y = \frac{1}{1 + x^2}$ en $y = \frac{x^2}{2}$. Gebruik verschillende kleuren voor de krommen.Bepaal de snijpunten en bereken de oppervlakte van het gebied ingesloten door beide krommen. Geef daarvoor een analytische en een numerieke waarde. Gebruik daarnaast ook een numerieke integratiemethode en vergelijk de resultaten.

```python
from sympy.plotting import plot
x = sp.symbols("x")
f1 = 1/(1+x**2)
f2 = (x**2)/2
vgl = sp.Eq(f1-f2,0)
xsnijpunt = sp.solve(vgl,x)
print(xsnijpunt)

x1=xsnijpunt[0]
print("x1 = ",x1)
x2=xsnijpunt[1]
print("x2 = ",x2)
y1=f1.subs(x,x1)
print("y1 = ",y1)
y2=f1.subs(x,x2)
print("y2 = ",y2)

plotf=plot(f1,(x,-2,2), show=False,line_color="r")
plotg=plot(f2,(x,-2,2), show=False,line_color="g")
plotf.extend(plotg)
plotf.show()
```

```python
opp=sp.integrate(f1-f2,(x,x1,x2))
print("opp = ",opp)
numopp=opp.evalf()
print("numerieke opp = ",opp)

from scipy.integrate import quad
def fnum(x):
    return 1/(1+x**2)-x**2/2
benadering,fout = quad(fnum,x1,x2)
print("benadering = ", benadering)
print("fout = ",fout)
```

### Oefening 26
Maak een 3D plot en een contourplot van $z = \frac{\sin \sqrt{x^2 + y^2}}{\sqrt{x^2 + y^2}}$, $x \in [-10, 10]$ en $y \in [-10, 10]$.

```python
xwaarden = np.linspace(-10,10,100)
ywaarden = np.linspace(-10,10,100)
[X,Y] = np.meshgrid(xwaarden,ywaarden)
Z = np.sin(np.sqrt(X**2+Y**2))/np.sqrt(X**2+Y**2)

fig1,ax1=plt.subplots()
ax1.contourf(X,Y,Z)
ax1.set_xlabel("X")
ax1.set_ylabel("Y")

fig2 = plt.figure()
ax2 = plt.axes(projection = "3d")
ax2.plot_surface(X,Y,Z) # type: ignore
```

### Oefening 30
Teken een kromme in het complexe vlak, beschreven door $z = t e^{jt}$, $t \in [0, 4\pi]$.

```python
t=np.linspace(0,4*np.pi,500)
z=t*np.exp(1j*t)

plt.plot(z.real,z.imag)
plt.xlabel("Re(z)")
plt.ylabel("Im(z)")
plt.axis("equal")
plt.show()
```

### Oefening 33
Bepaal de reële oplossingen van $\begin{cases} x^2 - 2x + y^2 = 0 \\ 9x^2 - 18x + 4y^2 = 27 \end{cases}$ 

Verklaar grafisch het resultaat.

```python
from sympy.plotting import plot

x,y = sp.symbols('x y')
vgl1 = sp.Eq(x**2-2*x+y**2,0)
vgl2 = sp.Eq(9*x**2-18*x+4*y**2,27)
opl = sp.solve((vgl1,vgl2),(x,y))
print(opl)

plt1 = sp.plotting.plot_implicit(vgl1,(x,-10,10),(y,-10,10), show=False,line_color='r') # type: ignore
plt2 = sp.plotting.plot_implicit(vgl2,(x,-10,10),(y,-10,10), show=False,line_color='g') # type: ignore
plt1.extend(plt2)
plt1.show()
```
